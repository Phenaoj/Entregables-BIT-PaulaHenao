```python
#Detección de valores sospechosos e incosistentes
for columna in df.select_dtypes(include="object").columns:
    print(f"\n--- {columna} ---")
    print(df[columna].value_counts(dropna=False))

#Problemas identificados: las columnas como Animal type, Country, Gender, Animal Name contienen valores vacios los cuales se deben analizar si son datos relevantes para tener en cuenta o solo harían ruido en el análisis,
#en la columna Animal type se encuentra una misma especie escrita de dos maneras diferentes,
#en la columna Country contamos con los nombres completos pero también con las abreviaciones.
```

```python
# 2. LIMPIEZA DE DATOS
df.head(6)
```

```python
# Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.
df[df["Animal type"].isna()]

# Se puede identificar que una gran mayoría de los registros que tienen valores faltantes en el Animal Type, no aportan información relevante ya que casi todas las columnas están vacías; por consiguiente se puede descartar dichos valores.
```

```python
df = df.dropna(subset=["Animal type"])

# Detectar y elimina los registros duplicados cuando corresponda.
# Se va a eliminar aquellas filas que sean completamente iguales
df = df.drop_duplicates()
```

```python
#Corrección de la columna Animal Type
#Para el caso de Ledgehod, Wedgehod y Hedgehog, se busca en internet y solo el úlitmo nombre arroja un resultado de especie de animales, por ende se estandarizará a dicha especie.
#Caso análogo con el European bison.
df["Animal type"] = df["Animal type"].replace({
    "red squirrell": "red squirrel",
    "red squirel": "red squirrel",
    "ledgehod": "hedgehog",
    "wedgehod": "hedgehog",
    "European bisson": "European bison",
    "European buster": "European bison",
    "European bison™": "European bison",
    "lynx?": "lynx"})

#Corrección de la columna Country
# Estándarizar las abreviaciones del país a su nombre
df["Country"] = df["Country"].replace({
    "PL": "Poland",
    "DE": "Germany",
    "HU": "Hungary",
    "Hungry": "Hungary",
    "CZ": "Czech Republic",
    "Czech": "Czech Republic"})
```

```python
# Estandarización del formato para la columan Observation Date, teniendo en cuenta que tiene formatos mixtos. Se valida al final que si esté en el formato correcto y no haya valores nulos.
df["Observation date"] = pd.to_datetime(
    df["Observation date"],
    format="mixed",
    dayfirst=True)
print(df["Observation date"].dtype)
df["Observation date"].isna().sum()
```

```python
# Ahora se revisa la calidad de datos para las columnas Weight kg y Body Length cm
df[["Weight kg", "Body Length cm"]].describe()

#Aparece que hay valores negativos, los cuales no tienen sentido para hablar de peso y longitud de un animal
```

```python
#Ubicaremos los datos negativos para el peso
df[df["Weight kg"] < 0]

#Vemos que el registro tiene más información de valor, por lo que no justifica eliminarlo, sino modificarlos, se hará en el punto de Numpy
```

```python
#Comprobación de la limpieza
df.shape

#Pasamos de 1011 filas a 825
```

```python
df.duplicated().sum()

#Ya no tenemos valores duplicados
```

```python
for columna in ["Animal type", "Country"]:
    print(f"\n--- {columna} ---")
    print(df[columna].value_counts(dropna=False))

#Ya en Animal Type no existe registros vacios, además se simplificaron la cantidad de animales, para Country ya no se encuentra las abreviaciones, solo los nombres completos de los paises.
```

```python
# 3. USO DE NUMPY
# Calculamos la mediana de Red Squirrel tomando sus valores mayores que 0
mediana_red_squirrel = np.median(
    df.loc[(df["Animal type"] == "red squirrel") & (df["Weight kg"] >= 0),"Weight kg"])

mediana_red_squirrel
print(mediana_red_squirrel)

# Vamos a reemplazar los valores negativos por la mediana para poder terminar de limpiar los registros
df["Weight kg"] = np.where((df["Animal type"] == "red squirrel") & (df["Weight kg"] < 0), mediana_red_squirrel, df["Weight kg"])

#Cómo se mencionó anteriormente, se identicó un valor erroneo en los registros para peso, sin embargo, al ser registros válidos se optó por reemplazar dichos valores negativos con la mediana específicamente de dicha especie.
```

```python
# 4. ANÁLISIS CON GROUPBY
#¿Cuál es el número de observaciones por tipo de animal?
observations_animal = (df.groupby("Animal type").size().sort_values(ascending=False))

observations_animal
```

```python
#¿Cuál es el promedio de una variable numérica según el tipo de animal?
average_weight_animal = (df.groupby("Animal type")["Weight kg"].mean().sort_values(ascending=False))

average_weight_animal
```

```python
#¿Cuál es el peso promedio de cada tipo de animal en cada país?
average_weight_country_animal = (
    df.groupby(["Country", "Animal type"])["Weight kg"].mean().sort_index())

average_weight_country_animal
```

```python
#¿Cuántos registros de cada tipo de animal hay en cada país?
observations_animal_country = (
    df.groupby(["Country", "Animal type"]).size().reset_index(name="Observations")
    .sort_values(["Country", "Observations"],ascending=[True, False]))

observations_animal_country
```

```python
#¿Cuántas observaciones de cada tipo de animal fueron registradas por cada persona que compiló los datos?
observations_compiler_animal = (
    df.groupby(["Data compiled by", "Animal type"]).size()
    .reset_index(name="Observations")
    .sort_values(["Data compiled by", "Observations"],ascending=[True, False]))

observations_compiler_animal
```

```python
#¿Cuántos registros tiene cada persona que compiló los datos en cada país?
observations_compiler_country = (
    df.groupby(["Country", "Data compiled by"]).size().reset_index(name="Observations")
    .sort_values(["Country", "Observations"],ascending=[True, False]))

observations_compiler_country
```
