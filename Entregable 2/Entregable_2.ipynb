```python
# 2. LIMPIEZA DE DATOS
df.head(6)
```

```python
# Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.
df[df["Animal type"].isna()]

# Se puede identificar que una gran mayoría de los registros que tienen valores faltantes en el Animal Type, no aportan información relevante ya que casi todas las columnas están vacías; por consiguiente se puede descartar dichos valores.
```

```python
df = df.dropna(subset=["Animal type"])

# Detectar y elimina los registros duplicados cuando corresponda.
# Se va a eliminar aquellas filas que sean completamente iguales
df = df.drop_duplicates()
```

```python
#Corrección de la columna Animal Type
#Para el caso de Ledgehod, Wedgehod y Hedgehog, se busca en internet y solo el úlitmo nombre arroja un resultado de especie de animales, por ende se estandarizará a dicha especie.
#Caso análogo con el European bison.
df["Animal type"] = df["Animal type"].replace({
    "red squirrell": "red squirrel",
    "red squirel": "red squirrel",
    "ledgehod": "hedgehog",
    "wedgehod": "hedgehog",
    "European bisson": "European bison",
    "European buster": "European bison",
    "European bison™": "European bison",
    "lynx?": "lynx"})

#Corrección de la columna Country
# Estándarizar las abreviaciones del país a su nombre
df["Country"] = df["Country"].replace({
    "PL": "Poland",
    "DE": "Germany",
    "HU": "Hungary",
    "Hungry": "Hungary",
    "CZ": "Czech Republic",
    "Czech": "Czech Republic"})
```

```python
#Estandarización del formato para la columan Observation Date, se verifica si después de los cambios aplicados aún existen fechas en otro formato
print(df["Observation date"].dtype)
df["Observation date"].isna().sum()

#Como ya se encuentra en el formato adecuado y no hay valores vacíos, no es necesario hacer nada más.
```

```python
for columna in ["Animal type", "Country"]:
    print(f"\n--- {columna} ---")
    print(df[columna].value_counts(dropna=False))

#Ya en Animal Type no existe registros vacios, además se simplificaron la cantidad de animales, para Country ya no se encuentra las abreviaciones, solo los nombres completos de los paises.
```
